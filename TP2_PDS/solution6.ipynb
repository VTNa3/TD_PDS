```python
grades = {
    "Dara":   [80, 75, 90],
    "Sophea": [95, 88, 92],
    "Vibol":  [40, 55, 45],
    "Malis":  [70, 85, 78],
    }
```

```python
averages = {}
for name, scores in grades.items():
    averages[name] = round(sum(scores) / len(scores), 1)
    print("Averages:", averages)
```

```python
passed = set() 
for name, avg in averages.items():
    if avg >= 70:
        print(name, "PASS")
        passed.add(name)
    else:
        print(name, "FAIL") 
        print("Passed:", sorted(passed))
```

```python
ranking = sorted(averages.items(), key=lambda item: item[1], reverse=True)
for rank, (name, avg) in enumerate(ranking, start=1):
    print(f"#{rank} {name} ({avg})")
```
