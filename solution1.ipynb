```python
# Step1
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
temps = [31, 33, 29, 34, 5, 332, 30]
```

```python
#step2
print("Number of days:", len(temps))
print("First day:", days[0],temps[0])
print("Last day:", days[-1], temps[-1])
```

```python
#step3
print("Hightest temps:", max(temps))
print("Lowest temps:",min(temps))
average = sum(temps) / len(temps)
print("Average:",round(average,1))
```

```python
#step4
hottest = temps.index(max(temps))
print("Hottest day:", days[hottest])
```

```python
#step5
print("sorted list: ", sorted(temps))
print("original:",temps)
```
